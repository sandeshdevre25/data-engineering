```python
import pandas as pd
import json
import xml.etree.ElementTree as ET
from bs4 import BeautifulSoup
from io import StringIO

# 1. Read Text File
with open("sample.txt", "w") as f:
    f.write("Hello Python\nData Processing")

with open("sample.txt", "r") as f:
    print("TEXT FILE:", f.read())

# 2. Read CSV File
csv_data = """Name,Age,Marks
Rahul,20,85
Priya,21,90
Amit,,75"""

df = pd.read_csv(StringIO(csv_data))
print("\nCSV DATA:\n", df)

# Check missing values
print("\nMissing Values:\n", df.isnull().sum())

# 3. Read HTML
html = "<table><tr><th>Name</th></tr><tr><td>Rahul</td></tr></table>"
print("\nHTML DATA:\n", pd.read_html(StringIO(html))[0])

# 4. Read XML
xml_data = "<students><student>Rahul</student></students>"
root = ET.fromstring(xml_data)
print("\nXML DATA:", root.find("student").text)

# 5. Read JSON
json_data = '{"name": "Rahul", "age": 20}'
data = json.loads(json_data)
print("\nJSON DATA:", data)

# Check duplicate records
print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
# Write binary file
data = bytes([65, 66, 67, 68, 69])

with open("sample.bin", "wb") as file:
    file.write(data)

print("Binary file written successfully.")

# Read binary file
with open("sample.bin", "rb") as file:
    content = file.read()

print("Binary data:", content)
print("Decoded data:", content.decode())
```

```python
import re

text = "My phone number is 9876543210. My city is Mumbai."

# Search for a phone number
result = re.search(r"\d{10}", text)
print("Phone Number:", result.group() if result else "Not found")

# Split the string
words = re.split(r"\s+", text)
print("Split Text:", words)

# Replace phone number
new_text = re.sub(r"\d{10}", "XXXXXXXXXX", text)
print("Replaced Text:", new_text)

# Find all words starting with M
matches = re.findall(r"\bM\w*", text)
print("Words starting with M:", matches)
```

```python
import sqlite3

# Connect to database
conn = sqlite3.connect("college.db")
cursor = conn.cursor()

# Create table
cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
    id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER,
    marks REAL
)
""")

# CREATE: Insert a record
cursor.execute(
    "INSERT OR REPLACE INTO students VALUES (?, ?, ?, ?)",
    (1, "Rahul", 20, 85)
)
conn.commit()
print("Record inserted successfully.")

# READ: Display records
cursor.execute("SELECT * FROM students")
print("Student Records:", cursor.fetchall())

# UPDATE: Change marks
cursor.execute(
    "UPDATE students SET marks = ? WHERE id = ?",
    (95, 1)
)
conn.commit()
print("Record updated successfully.")

# DELETE: Remove a record
cursor.execute("DELETE FROM students WHERE id = ?", (1,))
conn.commit()
print("Record deleted successfully.")

# Display remaining records
cursor.execute("SELECT * FROM students")
print("Final Records:", cursor.fetchall())

conn.close()
```
